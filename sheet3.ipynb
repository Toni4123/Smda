a)

counting of hits in a fixed time interval -> $\textbf{Poisson distribution}$

-> $P(k|\lambda) = \frac{\lambda^k}{k!} \cdot exp(-\lambda)$ 

k:observed number of events

$\lambda$: expected mean


b) background only hypothesis

-> $\lambda = b = 2.0$

$P(k \ge 8|\lambda =2) = 1- P(n \le 7)$

OR for k being natural numbers: $P(k \ge 8|\lambda =2) =P(k > 7|\lambda =2)$




```python
from scipy.stats import poisson

lambd = 2
p = poisson.sf(7, lambd) #survival fraction calculates k>7

print(p)
```

$P(k \ge 8|\lambda =2) = 0.001097$

c) 

use inverted survival fraction

```python
from scipy.stats import norm

Z = norm.isf(p)

print("Z =", Z)
```

Z = 3.0627
