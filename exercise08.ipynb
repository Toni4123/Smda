a)
Find normalization constant N:
$$
\int_{- \pi}^{\pi} N \cdot \mathrm{exp}(- \left| \Delta \Psi \right| \cdot k)
$$
Aus Symmetriegründen kann das Integral von $- \pi$ bis $\pi$ auch geschrieben werden als zweimal das Integral von 0 bis $\pi$. Daraus folgt:
$$
N = \frac{1}{2(-\frac{1}{k} \mathrm{exp}(- \Delta \Psi))\Big|_0^\pi} \\
N = \frac{k}{2(1- \mathrm{exp}(- \pi k))} \\
\rightarrow PDF = \frac{k}{2(1- \mathrm{exp}(-\pi k))} \cdot \mathrm{exp}((- \left| \Delta \Psi \right| k))
$$

b)
From exercise a) we have:
$$
PDF = f(\Delta \Psi)
=
N \exp(-k|\Delta \Psi|) \mathrm{with} N = \frac{k}{2\left(1-\exp(-\pi k)\right)}
$$
The CDF is given by
$$
F(\Delta \Psi)
=
\int_{-\pi}^{\Delta \Psi}
f(t)\,dt
$$
Since the PDF contains an absolute value, the calculation for the CDF must be split into two seperate cases. \\
Case 1: $\Delta \Psi < 0$
$$
f(t) = N \exp(kt) \\
\rightarrow F(\Delta \Psi) = \int_{-\pi}^{\Delta \Psi} N \exp(kt)\,dt = \frac{N}{k} \left[\exp(kt)\right]_{-\pi}^{\Delta \Psi} \\
\Leftrightarrow F(\Delta \Psi) = \frac{N}{k}\left(\exp(k\Delta \Psi)-\exp(-k\pi)\right)
$$
Since for $\Delta \Psi < 0$:
$$
|\Delta \Psi| = -\Delta \Psi \\
\Leftrightarrow \exp(k\Delta \Psi) = \exp(-k|\Delta \Psi|)
$$
Using N gives:
$$
\frac{1}{2}=\frac{N}{k}\left(1-\exp(-k\pi)\right) \\
\Leftrightarrow F(\Delta \Psi)=\frac{1}{2}-\frac{N}{k}\left(1-\exp(-k|\Delta \Psi|)\right)
$$

Case 2: $\Delta \Psi > 0$

$$
f(t) = N \exp(-kt) \\
\rightarrow F(\Delta \Psi) =
\int_{-\pi}^{0} N \exp(-k|t|)\,dt +
\int_{0}^{\Delta \Psi} N \exp(-kt)\,dt
$$

Since for $t \in [-\pi,0]$:
$$
|t| = -t \\
\Rightarrow f(t) = N \exp(kt)
$$

First part:
$$
\int_{-\pi}^{0} N \exp(kt)\,dt =
\frac{N}{k}\left[\exp(kt)\right]_{-\pi}^{0} \\
\Leftrightarrow =
\frac{N}{k}\left(1-\exp(-k\pi)\right)
$$

Second part:
$$
\int_{0}^{\Delta \Psi} N \exp(-kt)\,dt =
\frac{N}{k}\left[-\exp(-kt)\right]_{0}^{\Delta \Psi} \\
\Leftrightarrow =
\frac{N}{k}\left(1-\exp(-k\Delta \Psi)\right)
$$

Adding both parts:
$$
F(\Delta \Psi)=
\frac{N}{k}\left(1-\exp(-k\pi)\right)+
\frac{N}{k}\left(1-\exp(-k\Delta \Psi)\right)
$$

Using normalization:
$$
\frac{1}{2}=\frac{N}{k}\left(1-\exp(-k\pi)\right)
$$

Substituting:
$$
\Leftrightarrow
F(\Delta \Psi)=
\frac{1}{2}+
\frac{N}{k}\left(1-\exp(-k\Delta \Psi)\right)
$$

Since for $\Delta \Psi > 0$:
$$
|\Delta \Psi| = \Delta \Psi \\
\Leftrightarrow
F(\Delta \Psi)=
\frac{1}{2}+
\frac{N}{k}\left(1-\exp(-k|\Delta \Psi|)\right)
$$
Combining both cases with sign:
$$
F(\Delta \Psi)=
\frac{1}{2}+
\operatorname{sign}(\Delta \Psi)\frac{N}{k}\left(1-\exp(-k|\Delta \Psi|)\right)
$$

c) 
Inverse Cumulative Distribution (PPF)
Start with the CDF from exercise b):
$$
F(x)=\frac{1}{2}+\frac{N}{k}\,\operatorname{sign}(x)\left(1-\exp(-k|x|)\right)
$$
Define:$q = F(x), \quad q \in (0,1)$ \\

Case 1: $q > 0.5$
$$
q = \frac{1}{2} + \frac{N}{k}\left(1-\exp(-kx)\right) \\
\Leftrightarrow q - \frac{1}{2} = \frac{N}{k}\left(1-\exp(-kx)\right) \\
\Leftrightarrow \frac{k}{N}\left(q - \frac{1}{2}\right) = 1 - \exp(-kx) \\
\Leftrightarrow \exp(-kx) = 1 - \frac{k}{N}\left(q - \frac{1}{2}\right) \\
\Leftrightarrow -kx = \ln\left(1 - \frac{k}{N}\left(q - \frac{1}{2}\right)\right) \\
\Leftrightarrow x = -\frac{1}{k}\ln\left(1 - \frac{k}{N}\left(q - \frac{1}{2}\right)\right)
$$

Case 2: $q < 0.5$
$$
q = \frac{1}{2} - \frac{N}{k}\left(1-\exp(-k|x|)\right) \\
\Leftrightarrow \frac{1}{2} - q = \frac{N}{k}\left(1-\exp(-k|x|)\right) \\
\Leftrightarrow \frac{k}{N}\left(\frac{1}{2} - q\right) = 1 - \exp(-k|x|) \\
\Leftrightarrow \exp(-k|x|) = 1 - \frac{k}{N}\left(\frac{1}{2} - q\right) \\
\Leftrightarrow -k|x| = \ln\left(1 - \frac{k}{N}\left(\frac{1}{2} - q\right)\right) \\
\Leftrightarrow |x| = -\frac{1}{k}\ln\left(1 - \frac{k}{N}\left(\frac{1}{2} - q\right)\right)
$$

Since $q < 0.5 \Rightarrow x < 0$:

$$
\Leftrightarrow x = \frac{1}{k}\ln\left(1 - \frac{k}{N}\left(\frac{1}{2} - q\right)\right)
$$

Compact Form:
$$
x =
\operatorname{sign}\left(q-\frac{1}{2}\right)
\cdot
\frac{1}{k}
\ln\left(1 - \frac{k}{N}\left|q-\frac{1}{2}\right|\right)^{-1}
$$
