Exercise 2: Dice

Total number of outcomes for a dice with 6 (equally probable) sides:
$$|\Omega| = 36$$
(a)
6 different possible outcomes:
(1,6)
(6,1)
(2,5)
(5,2)
(3,4)
(4,3)
$$P(W_{\text{red}} + W_{\text{blue}} = 7) = \frac{6}{36}$$

(b)
For a sum of pips $\geq 8$ the possible sums are: $8,9,10,11,12$ so the number of possible outcomes are: $5 + 4 + 3 + 2 + 1 = 15$:
$$P(W_{\text{red}} + W_{\text{blue}} \geq 8) = \frac{15}{36} = \frac{5}{12}$$

(c)
The possible outcomes are: (3,6) and (6,3)
$$P((W_{\text{red}} = 3 \cap W_{\text{blue}} = 6) \cup (W_{\text{red}} = 6 \cap W_{\text{blue}} = 3)) = \frac{2}{36} = \frac{1}{18}$$
(d)
There is only one possible outcome: (1,2)
$$P(W_{\text{red}} = 1 \land W_{\text{blue}} = 2) = \frac{1}{36}$$

(e)
When the red dice is a 3, the blue dice needs to be 5 to get a sum of 8
$$P(W_{\text{blue}} + W_{\text{red}} = 8 | W_{\text{red}} = 3) = \frac{1}{6}$$
(f)
When the red dice is a 3, the blue dice needs to be a 1 or larger, which is always true
$$P(W_{\text{blue}} + W_{\text{red}} ≥ 4 | W_{\text{red}} = 3) = \frac{6}{6} = 1$$

(g)
We already know, that the red dice shows a 3, so only the blue one matters
$$P(W_{\text{blue}} = 2 | W_{\text{red}} = 3) = \frac{1}{6}$$
